<font color="red">**2. ¿Qué diferencia existe entre Feature Selection y Feature Engineering? Proporciona un ejemplo de cada uno.** </font>

Cada una resuelve problematicas muy puntuales: 

**Feature Selection:** Consiste en la eleccion adecuada de variables evitando caer en sesgos de selección o en variables espuria. 

Un ejemplo seria: Para un modelo de prediccion de venta de helado en plazas publicas la variable de numero de personas en la plaza por dia seria una variable importante ignorando variables como numero de patentes registradas en el IMPI ese dia, ya que no ayudan en nada al modelo y no tiene una relacion de causalidad o correlacion directa con la variable de importancia.


**Feature Engineering:** Consiste en la mejor adopcion de estas variables para su uso.
Un ejemplo seria: Retomando, el numero de personas como una variable importante lo seria aun mas el numero de personas por metro cuadrado en cada plaza. Convirtiendo a la densidad poblacional en una variable aun mejor para el modelo.

<font color="red">**3. Selecciona alguno de los siguientes modelos y explica de forma cómo funciona el algoritmo**</font>
(Random Forest | Support Vector Machines | Regression Multiple)

**Regression Multiple:**

En su forma base:
$$Y = \beta_0 + \beta_1 X_1 + \beta_2 X_2 + \dots + \beta_n X_n + \varepsilon$$

Tienen la finalidad de hacer predicciones de una variable continua $Y$ mediante su relacion causal con otras variables $X_1,X_2...X_n$. Mas que un algoritmo es el modelo estadistico estandar, la calibracion de los parametros $\beta_0,\beta_2...\beta_n$ se da mediante el principio de Minimos Cuadrados Ordinarios (MCO)<sup><a href="#fn2" id="ref2">[2]</a></sup> que refiere a la que la suma de los residuos al cuadrado del diferencial de $Y-(\beta_0 + \beta_1 X_1 + \beta_2 X_2 + \dots + \beta_n X_n)$ en cada periodo $t_1,t_2...t_n$ debe ser el minimo posible. Esto asegura que el ajuste es el mejor posible a realizar y proporciona a las $\beta_n$ un ponderado explicativo a las variables $X_n$ siempre y cuando se cumplan los principios de no multicolinealidad, normalidad en los residuos, homocedasticidad e independencia en los residuos. 

El modelo base, puede mutar tambien en un sin fin de modelos lo que lo hace muy versatil ya sea para series temporales, de corte transversal o de panel de datos

**Pequeña mencion a los otros modelos**

**Random Forest**: Se genera mediante el conjunto de arboles de decision en donde se iteran arboles con sub conjuntos de datos de forma que cada arbol funciona como un generador de decision y la decision final esta dada por el voto mayoritario de los arboles en conjunto en caso de clasificacion y en regresiones por la estimacion promedio.

**Support Vector Machines:** Es un algoritmo principalmente ocupado en clasificacion que busca encontrar fronteras de decision en conjuntos de datos mediante MMC (Maximum Margin Classifier) que es la mayor distancia entre los puntos mas cercanos de cada clase. Su principal caracteristica es que puede trabajar bajo un sentido de correspondencia no lineal utilizando el principio de kernel.

<font color="red">**4. ¿Cómo reducen el error los métodos de Bagging (ej. Random Forest) y Boosting (ej. XGBoost) en términos de sesgo y varianza?**</font>

**El bagging** es la tecnica utilizada para reducir la varianza en la estimacion, es la principal caracteritica de los random forest ya se centra en repeticion de multiples modelos mas pequeños con submuentras de informacion para una toma de decisiones conjuta.

$$\text{Varianza del ejercicio conjunto} = \rho \sigma^2 + \frac{1 - \rho}{N} \sigma^2$$

Esto es devido a que la varianza total del ejercicio esta en funcion de la varianza individual de cada ejercicio y la correlacion media de manera que si aumentamos el numero de ejercicios la varianza individual de la estimacion disminuye en factor de ajuste $1-\rho$.


**El Boosting** es la tecnica que se utiliza como funcion de aprendizaje ya que se busca reducir la estimacion de error de los modelos anteriores a traves de aprendizaje secuencial. Esto mitiga el sesgo, no obstante el uso excesivo crea complejidad inecesaria y sobreajute ya que el modelo deja de generalizar y se apega a buscar encajar con todos los datos.

Ambas tecnicas deben ser usadas en su conjunto para corregir el error de estimacion:

$$\text{Error Total} = \text{Sesgo}^2 + \text{Varianza} + \text{Error Irreducible}$$

<font color="red">**7. Explica con tus propias palabras los conceptos de sobreajuste (Overfitting) y subajuste (Underfitting). ¿Cómo podrías detectar que tu modelo está sufriendo de sobreajuste durante el entrenamiento? Menciona al menos dos técnicas para mitigarlo.**</font>

Respuesta: 


El sobreajuste (overfitting) ocurre cuando un modelo de machine learning aprende de forma excesiva los patrones específicos y el ruido de los datos de entrenamiento, lo que arruina su capacidad para generalizar con datos nuevos. Por el contrario, el subajuste (underfitting) se presenta cuando un modelo es demasiado simple para capturar la estructura subyacente de los datos, resultando en un rendimiento pobre tanto en el conjunto de entrenamiento como en el de validación.

La señal principal se identifica al monitorear las curvas de aprendizaje (pérdida o métrica de rendimiento frente a las épocas). El sobreajuste se manifiesta cuando el error en el conjunto de entrenamiento continúa disminuyendo de manera constante, mientras que el error en el conjunto de validación alcanza un punto mínimo y comienza a incrementarse de nuevo. Esta divergencia indica que el algoritmo ha dejado de aprender reglas generales y ha comenzado a memorizar las muestras específicas de entrenamiento.

<font color="red">**10. Describe las principales diferencias entre las responsabilidades de un Científico de Datos y un Ingeniero de Machine Learning. Posteriormente, explica cómo debería realizarse el proceso de Model Handoff para asegurar una transición exitosa del modelo hacia producción**</font>

Respuesta:

El Científico de Datos se enfoca en la exploración, la estadística y el descubrimiento de patrones. Su objetivo principal es formular hipótesis, realizar análisis exploratorio (EDA) y construir prototipos de modelos en entornos de experimentación (como Jupyter Notebooks) para responder preguntas de negocio y validar su rendimiento analítico.

El Ingeniero de Machine Learning, en cambio, se enfoca en la ingeniería de software, la arquitectura y la producción. Su objetivo es tomar los prototipos funcionales del Científico de Datos, refactorizar su código y llevarlos a entornos productivos mediante pipelines escalables, APIs de baja latencia, contenedores e infraestructura automatizada (MLOps), garantizando la disponibilidad y el rendimiento del sistema en tiempo real.

Proceso de Model Handoff
1.1. Modularización y Serialización:Preparación de artefactos.Se convierte el código de scripts/notebooks a módulos reutilizables y se exportan los artefactos del modelo junto con sus transformadores (escaladores, encoders) en formatos estándar como ONNX, Pickle o Joblib.

2.2. Definición del Contrato de Inferencia:Estandarización de entradas y salidas.Data Scientists y ML Engineers fijan el esquema de datos que consumirá y devolverá el modelo (mediante JSON Schemas o clases Pydantic), definiendo además la modalidad de servicio (Real-time REST API o Batch).

3.3. Pruebas de Alineación y Carga:Prevención del Training-Serving Skew.Se ejecutan pruebas unitarias e integradas para garantizar que las transformaciones en producción sean idénticas a las de entrenamiento, verificando también que la latencia y el uso de memoria cumplan con los SLAs de infraestructura.

4.4. Despliegue Progresivo y Monitoreo:Paso a producción y observabilidad.El modelo se integra en un pipeline de CI/CD para ser desplegado gradualmente (mediante estrategias Canary o Shadow deployment) y se configuran alertas para detectar degradación operativa y desviaciones de datos (Data Drift).
